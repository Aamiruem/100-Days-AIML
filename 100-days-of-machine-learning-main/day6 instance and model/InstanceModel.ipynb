```python
from sklearn.neighbors import KNeighborsRegressor

x = [[500], [1000], [1500], [2000]]
y = [20, 40, 60, 80]

model = KNeighborsRegressor(n_neighbors=2)
model.fit(x, y)

print(model.predict([[1200]]))
```

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.neighbors import KNeighborsRegressor

x = [[500], [1000], [1500], [2000]]
y = [20, 40, 60, 80]

model = KNeighborsRegressor(n_neighbors=2)
model.fit(x, y)

x_line = np.linspace(400, 2100, 300).reshape(-1, 1)
plt.scatter(x, y, label="Actual data")
plt.plot(x_line, model.predict(x_line), color="red", label="KNN prediction (k=2)")
plt.scatter(
    [[1200]],
    model.predict([[1200]]),
    color="green",
    marker="x",
    s=100,
    label="Prediction at 1200",
)
plt.xlabel("Area")
plt.ylabel("Target Value")
plt.title("KNN Regression Graph")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

x = [[500], [1000], [1500], [2000]]
y = [20, 40, 60, 80]

model = LinearRegression()
model.fit(x, y)

predictions = model.predict(x)

plt.scatter(x, y, label="Actual data")
plt.plot(x, predictions, color="red", label="Regression line")
plt.xlabel("Area")
plt.ylabel("Target value")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)
plt.show()
print(model.predict([[1000]]))
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
import matplotlib.pyplot as plt

x = [[500], [1000], [1500], [2000]]
y = [20, 40, 60, 80]

model = LinearRegression()
model.fit(x, y)

pred = model.predict(x)
print("R² Score:", r2_score(y, pred))  # 1.0

plt.scatter(x, y, label="Actual")
plt.plot(x, pred, color="red", label="Predicted")
plt.legend()
plt.grid(True)
plt.show()

print("Prediction:", model.predict([[1200]]))  # [48.]
```
